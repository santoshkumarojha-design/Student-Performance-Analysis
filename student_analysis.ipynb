```python
# Student Performance Data Analysis
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("student-mat.csv", sep=";")

# Display first 5 rows
df.head()
```

```python
# Dataset Shape
df.shape
```

```python
# Data Types
df.dtypes
```

```python
# Missing Values
df.isnull().sum()
```

```python
## Duplicate Records
print("Duplicates before:", df.duplicated().sum())
df = df.drop_duplicates()
print("Duplicates after:", df.duplicated().sum())
```

```python
# Average Final Grade
avg_grade = df["G3"].mean()
print("Average Final Grade:", avg_grade)
```

```python
# Students Scoring Above 15
above_15 = (df["G3"] > 15).sum()
print("Students scoring above 15:", above_15)
```

```python
# Study Time vs Performance
corr = df["studytime"].corr(df["G3"])
print("Correlation:", corr)
```

```python
if corr > 0:
    print("Positive correlation")
elif corr < 0:
    print("Negative correlation")
else:
    print("No correlation")
```

```python
# Gender-wise Performance
gender_performance = df.groupby("sex")["G3"].mean()
print(gender_performance)
```

```python
# Distribution of Final Grades
plt.hist(df["G3"], bins=10, edgecolor='black', linewidth=0.5)
plt.title("Distribution of Final Grades")
plt.xlabel("Grade")
plt.ylabel("Number of Students")
plt.show()
```

```python
# Study Time vs Final Grade
plt.scatter(x=df["studytime"], y=df["G3"])
plt.title("Study Time vs Final Grade")
plt.xlabel("Study Time")
plt.ylabel("Final Grade")
plt.show()
```

```python
# Gender-wise Average Final Grade
gender_data = df.groupby("sex")
gender_avg = gender_data["G3"].mean()

plt.bar(gender_avg.index, gender_avg.values)
plt.title("Average Grade by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Grade")

plt.show()
```
